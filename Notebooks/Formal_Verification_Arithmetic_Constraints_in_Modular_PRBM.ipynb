3. **Graph Sparsity Base Block (`mask_base_block_density`):** Proves that the arithmetic mask allows exactly 2 valid transitions in every sequence of 6 lattice sites. This is the discrete topological foundation for the expected degree asymptotic limit in Lemma 1.
4. **Bipartite Factorization Bound (`bipartite_conjecture_bound`):** Certifies the tighter conjectured bound ($D_2 \le 1/4$). It proves that if the topological space divides symmetrically into the two active chiral sublattices ($\mathcal{C}_1$ and $\mathcal{C}_5$), the maximum factorized probability density evaluates exactly to $0.25$.
