### **Physical & Mathematical Context**

In power-law random banded matrix (PRBM) Hamiltonian models with sparse hopping constraints, systematic comparison across alternative masks ($\mathbb{Z}/6\mathbb{Z}$, Möbius $\mu(d) \neq 0$, Liouville $\lambda(d) = -1$, and Bernoulli random controls) reveals that the emergent Non-Ergodic Extended (NEE) phase is **entirely governed by the preservation of short-range (first-neighbor $d=1$) hopping**, rather than by the deep arithmetic structure of the mask.

While random masks with channel density $\rho = 1/3$ collapse to $D_2 \approx 0.06$ when $d=1$ is absent, forcing $d=1$ restores $D_2 \approx 0.28$, matching the coprime-to-6 mask. Removing $d=1$ from the coprime-to-6 mask collapses $D_2$ to $0.0448$. Furthermore, the Möbius mask is statistically indistinguishable from a random mask matching its first four short-range channels $d \in \{1, 2, 3, 5\}$ (Kolmogorov--Smirnov $p = 0.35$).

This companion interactive kernel formally certifies in **Lean 4** the core theoretical bounds, channel density hierarchies, and Kato--Rellich stability conditions governing the sparse-hopping PRBM interaction graph.
